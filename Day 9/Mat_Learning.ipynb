```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [21, 34, 22, 11, 55]
z = [26, 75, 44, 11, 19]

width = 0.1

plt.bar([i - width/2 for i in x], y, width=width, label='Y')
plt.bar([i + width/2 for i in x], z, width=width, label='Z')

plt.title('X vs Y vs Z')
plt.xlabel('X')
plt.ylabel('Values')
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Original values
original = np.array([1, 9, 7, 9, 10, 9, 4, 7, 2, 8])

# Calculations
x = np.sqrt(original)   # Square root
y = original ** 2       # Square
z = original * 2        # Double
w = original ** 3       # Cube

# Create 2x2 subplots
plt.figure(figsize=(12, 6))

# 1. Square Root
plt.subplot(2, 2, 1)
plt.plot(original, x, marker='o')
plt.title('Square Root')
plt.xlabel('Original')
plt.ylabel('Square Root')

# 2. Square
plt.subplot(2, 2, 2)
plt.plot(original, y, marker='o')
plt.title('Square')
plt.xlabel('Original')
plt.ylabel('Square')

# 3. Double
plt.subplot(2, 2, 3)
plt.plot(original, z, marker='o')
plt.title('Double')
plt.xlabel('Original')
plt.ylabel('Double')

# 4. Cube
plt.subplot(2, 2, 4)
plt.plot(original, w, marker='o')
plt.title('Cube')
plt.xlabel('Original')
plt.ylabel('Cube')

# Adjust spacing
plt.tight_layout()

plt.suptitle('All Four Charta Are Here')
# Display
plt.show()
```
