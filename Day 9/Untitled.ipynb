```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [21, 34, 22, 11, 55]
z = [26, 75, 44, 11, 19]

width = 0.1

plt.bar([i - width/2 for i in x], y, width=width, label='Y')
plt.bar([i + width/2 for i in x], z, width=width, label='Z')

plt.title('X vs Y vs Z')
plt.xlabel('X')
plt.ylabel('Values')
plt.legend()

plt.show()
```
