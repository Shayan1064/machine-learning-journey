```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.DataFrame({
    "Month": [
        "Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
    ],

    "Sales": [
        42000, 45000, 48000, 52000, 58000, 61000,
        65000, 63000, 59000, 68000, 72000, 78000
    ],

    "Profit": [
        8000, 9000, 9500, 11000, 13000, 14500,
        16000, 15000, 13500, 17500, 19000, 22000
    ],

    "Customers": [
        120, 135, 142, 155, 170, 185,
        200, 195, 180, 210, 225, 240
    ],

    "Marketing_Spend": [
        5000, 5200, 5500, 6000, 6500, 7000,
        7500, 7200, 6800, 8000, 8500, 9000
    ],

    "Orders": [
        95, 105, 110, 125, 140, 150,
        165, 160, 145, 175, 190, 205
    ]
})

print(df)
```

```python
df.head()
```

```python
plt.plot(df['Month'],df['Sales'],'o--r')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales in (RS)')
plt.grid()
plt.show()
# plt.style.avalible()
```

```python
plt.plot(df['Month'],'o--r')
plt.plot(df['Sales'],'o--g')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales in (RS)')
plt.grid()
plt.show()
```

```python
plt.bar(df['Month'],df['Sales'],color='green')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales in (RS)')
# plt.grid()
plt.show()
```

```python
plt.bar(df['Month'],color='green')
plt.bar(df['Sales'],color='orange')
plt.show()
```

```python
plt.hist(df['Sales'],color='skyblue',edgecolor='black',bins=12)
plt.show()
```

```python
plt.hist(df['Profit'],color='yellow',edgecolor='black',bins=12)
plt.show()
```

```python
plt.hist(df['Orders'],color='green',edgecolor='black',bins=12)
plt.show()
```
