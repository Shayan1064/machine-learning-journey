```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = iris.data
y = iris.target
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,2],c=y)
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage, dendrogram
```

```python
linked = linkage(X_scaled, method="ward")
```

```python
dendrogram(linked)
```

```python
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

# Create linkage
linked = linkage(X_scaled, method="ward")

# Plot dendrogram
plt.figure(figsize=(14, 7))

dendrogram(
    linked,
    truncate_mode="lastp",
    p=30,
    leaf_rotation=90,
    leaf_font_size=10,
    show_contracted=True
)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Cluster / Data Points")
plt.ylabel("Distance")
plt.axhline(y=7, linestyle="--")
plt.tight_layout()
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
model = AgglomerativeClustering(
    n_clusters=3
)

labels = model.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,2],c=labels)
```
